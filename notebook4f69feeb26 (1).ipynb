````python

# ============================================
# PLANT DISEASE CLASSIFICATION - COMPLETE CODE
# One MobileNetV2 model for all 15 classes
# ============================================

!pip -q install kagglehub

import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import kagglehub

from tensorflow.keras import layers, Model
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from sklearn.metrics import classification_report, ConfusionMatrixDisplay

# 1. DOWNLOAD DATASET
path = kagglehub.dataset_download("arjuntejaswi/plant-village")

# Automatically find the folder containing all disease classes
data_dir = os.path.join(path, "PlantVillage")

if not os.path.isdir(data_dir):
    raise FileNotFoundError(f"Dataset folder not found: {data_dir}")

class_names = sorted([
    d for d in os.listdir(data_dir)
    if os.path.isdir(os.path.join(data_dir, d))
])

assert len(class_names) == 15, (
    f"Expected 15 classes, found {len(class_names)}: {class_names}"
)

print("Dataset:", data_dir)
print("Number of classes:", len(class_names))
print(class_names)

# 2. LOAD DATA

import os
import numpy as np
import tensorflow as tf
from sklearn.model_selection import train_test_split
from collections import Counter

# -------------------------------
# 1. Configuration
# -------------------------------
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

# Supported image formats
IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp")

# -------------------------------
# 2. Read all classes and images
# -------------------------------
class_names = sorted([
    name for name in os.listdir(data_dir)
    if os.path.isdir(os.path.join(data_dir, name))
])

assert len(class_names) == 15, (
    f"Expected 15 classes, found {len(class_names)}"
)

image_paths = []
image_labels = []

for label, class_name in enumerate(class_names):
    class_dir = os.path.join(data_dir, class_name)

    files = sorted(os.listdir(class_dir))
    valid_files = [
        f for f in files
        if f.lower().endswith(IMAGE_EXTENSIONS)
        and os.path.isfile(os.path.join(class_dir, f))
    ]

    for filename in valid_files:
        image_paths.append(os.path.join(class_dir, filename))
        image_labels.append(label)

image_paths = np.array(image_paths)
image_labels = np.array(image_labels, dtype=np.int32)

print("Total images:", len(image_paths))
print("Total classes:", len(class_names))

# -------------------------------
# 3. Stratified train/test split
#    70% train, 30% temporary
# -------------------------------

import re
from sklearn.model_selection import StratifiedGroupKFold

# 1. groups banao (har image ka leaf ID)
def group_key(path):
    name = os.path.splitext(os.path.basename(path))[0]
    return re.sub(r'_(flip\w*|new\w*|\d+deg\w*)$', '', name, flags=re.I)

groups = np.array([group_key(p) for p in image_paths])

# 2. split (yahan groups use hue)
sgkf = StratifiedGroupKFold(n_splits=7, shuffle=True, random_state=SEED)
tv_idx, test_idx = next(sgkf.split(image_paths, image_labels, groups))

sgkf2 = StratifiedGroupKFold(n_splits=6, shuffle=True, random_state=SEED)
tr, va = next(sgkf2.split(image_paths[tv_idx], image_labels[tv_idx], groups[tv_idx]))
train_idx, val_idx = tv_idx[tr], tv_idx[va]

# 3. ab wahi purane variables bana do
train_paths, train_labels = image_paths[train_idx], image_labels[train_idx]
val_paths,   val_labels   = image_paths[val_idx],   image_labels[val_idx]
test_paths,  test_labels  = image_paths[test_idx],  image_labels[test_idx]

# 4. check
assert not set(groups[train_idx]) & set(groups[test_idx])
assert not set(groups[train_idx]) & set(groups[val_idx])
assert not set(groups[val_idx])   & set(groups[test_idx])



print("\nSplit sizes:")
print(f"Train:      {len(train_paths)} ({len(train_paths)/len(image_paths):.1%})")
print(f"Validation: {len(val_paths)} ({len(val_paths)/len(image_paths):.1%})")
print(f"Test:       {len(test_paths)} ({len(test_paths)/len(image_paths):.1%})")

# -------------------------------
# 5. Check per-class distribution
# -------------------------------
print("\nPer-class image counts:")

for i, name in enumerate(class_names):
    print(
        f"{i:2d} {name:48s}"
        f" Train={np.sum(train_labels == i):4d}"
        f" Val={np.sum(val_labels == i):4d}"
        f" Test={np.sum(test_labels == i):4d}"
    )

# -------------------------------
# 6. Image loading function
# -------------------------------
def load_image(path, label):
    image = tf.io.read_file(path)
    image = tf.io.decode_image(
        image,
        channels=3,
        expand_animations=False
    )
    image.set_shape([None, None, 3])
    image = tf.image.resize(image, IMG_SIZE)

    # Keep pixels in [0, 255].
    # MobileNetV2 preprocessing can remain inside your model.
    return image, label

# -------------------------------
# 7. Build datasets WITHOUT shuffle
# -------------------------------
def make_dataset(paths, labels, training=False):
    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    if training:
        # shuffle paths (cheap) BEFORE decoding images
        ds = ds.shuffle(len(paths), seed=SEED, reshuffle_each_iteration=True)
    ds = ds.map(load_image, num_parallel_calls=tf.data.AUTOTUNE)
    ds = ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
    return ds

train_ds = make_dataset(train_paths, train_labels, training=True)
val_ds   = make_dataset(val_paths,   val_labels)
test_ds  = make_dataset(test_paths,  test_labels)

print("\nDatasets created successfully!")
print("Train batches:", tf.data.experimental.cardinality(train_ds).numpy())
print("Validation batches:", tf.data.experimental.cardinality(val_ds).numpy())
print("Test batches:", tf.data.experimental.cardinality(test_ds).numpy())




# class_names = train_ds.class_names
NUM_CLASSES = len(class_names)

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)

# 3. DATA AUGMENTATION
augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.08),
    layers.RandomZoom(0.10),
])

# 4. LOAD PRETRAINED MOBILENETV2
base_model = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

base_model.trainable = False

inputs = layers.Input(shape=(224, 224, 3))

x = augmentation(inputs)
x = preprocess_input(x)

x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.30)(x)
outputs = layers.Dense(
    NUM_CLASSES,
    activation="softmax"
)(x)

model = Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# 5. TRAIN CLASSIFICATION HEAD
callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True
    ),
    tf.keras.callbacks.ModelCheckpoint(
        "/content/best_plant_model.keras",
        monitor="val_loss",
        save_best_only=True
    )
]



# 6. FINE-TUNE LAST LAYERS OF MOBILENETV2
base_model.trainable = True

# Freeze most layers; fine-tune only the last 30
for layer in base_model.layers[:-30]:
    layer.trainable = False

# Keep BatchNormalization layers frozen for stable fine-tuning
for layer in base_model.layers:
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history2 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=20,
    callbacks=callbacks
)

# 7. LOAD BEST SAVED MODEL
model = tf.keras.models.load_model(
    "/content/best_plant_model.keras"
)

# 8. EVALUATE
val_loss, val_accuracy = model.evaluate(val_ds, verbose=1)

print(f"\nValidation loss: {val_loss:.4f}")
print(f"Validation accuracy: {val_accuracy * 100:.2f}%")

# 9. PLOT TRAINING CURVES
acc = history2.history["accuracy"]
val_acc = (
history2.history["val_accuracy"]
)

loss =  history2.history["loss"]
val_loss_history = (
     history2.history["val_loss"]
)

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(acc, label="Training accuracy")
plt.plot(val_acc, label="Validation accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.title("Accuracy")

plt.subplot(1, 2, 2)
plt.plot(loss, label="Training loss")
plt.plot(val_loss_history, label="Validation loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.title("Loss")

plt.tight_layout()
plt.show()

# 10. CLASSIFICATION REPORT
y_true = np.concatenate([
    labels.numpy() for _, labels in val_ds
])

y_prob = model.predict(val_ds, verbose=1)
y_pred = np.argmax(y_prob, axis=1)

print(classification_report(
    y_true,
    y_pred,
    labels=np.arange(NUM_CLASSES),
    target_names=class_names,
    zero_division=0
))

# 11. CONFUSION MATRIX
plt.figure(figsize=(15, 15))

ConfusionMatrixDisplay.from_predictions(
    y_true,
    y_pred,
    labels=np.arange(NUM_CLASSES),
    display_labels=class_names,
    xticks_rotation=90,
    cmap="Blues",
    normalize="true",
    values_format=".2f"
)

plt.title("Normalized Confusion Matrix")
plt.tight_layout()
plt.show()

# 12. SAVE FINAL MODEL AND CLASS LABELS
model.save("/content/plant_disease_model.keras")

with open("/content/plant_class_names.txt", "w") as f:
    f.write("\n".join(class_names))

print("Model saved: /content/plant_disease_model.keras")
print("Class labels saved: /content/plant_class_names.txt")

# 13. TEST ON ONE VALIDATION IMAGE
for images, labels in val_ds.take(1):
    image = images[0:1]
    actual_label = int(labels[0].numpy())

    probabilities = model.predict(image, verbose=0)[0]
    predicted_label = int(np.argmax(probabilities))

    plt.figure(figsize=(5, 5))
    plt.imshow(images[0].numpy().astype("uint8"))
    plt.axis("off")
    plt.title(
        f"Predicted: {class_names[predicted_label]}\n"
        f"Confidence: {probabilities[predicted_label] * 100:.2f}%\n"
        f"Actual: {class_names[actual_label]}"
    )
    plt.show()

    print("Top 3 predictions:")
    for idx in np.argsort(probabilities)[-3:][::-1]:
        print(f"{class_names[idx]}: {probabilities[idx] * 100:.2f}%")
    break
````

````python
import keras
from keras import layers
from keras.optimizers import Adamax
from keras.callbacks import EarlyStopping


data_augmentation = keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
    layers.RandomTranslation(
        height_factor=0.1,
        width_factor=0.1
    ),
    layers.RandomContrast(0.1),
], name="data_augmentation")

inputs = keras.Input(shape=(224, 224, 3))

# Augmentation: active during training only
x = data_augmentation(inputs)

# Pretrained EfficientNetB0
base_model2 = keras.applications.EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(224, 224, 3)
)

base_model2.trainable = False

x = base_model2(x, training=False)

x = layers.GlobalAveragePooling2D()(x)
x = layers.BatchNormalization()(x)
x = layers.Dense(128, activation="relu")(x)
x = layers.Dropout(0.3)(x)

outputs = layers.Dense(15, activation="softmax")(x)

model2 = keras.Model(inputs, outputs)

model2.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


model2.summary()
````

````python
import matplotlib.cm as cm

def gradcam(model, base, img):                       # img: (224,224,3), 0-255
    x = tf.convert_to_tensor(img[None], tf.float32)
    head = model.layers[model.layers.index(base)+1:]
    with tf.GradientTape() as tape:
        fmap = base(x, training=False)               # (1,7,7,1280)
        tape.watch(fmap)
        y = fmap
        for l in head:
            y = l(y, training=False)
        cls = tf.argmax(y[0])
        score = y[:, cls]
    grads = tape.gradient(score, fmap)
    w = tf.reduce_mean(grads, axis=(1, 2))           # (1,1280)
    cam = tf.reduce_sum(fmap * w[:, None, None, :], axis=-1)[0]
    cam = tf.nn.relu(cam)
    cam = (cam / (tf.reduce_max(cam) + 1e-8)).numpy()
    return cam, int(cls), y[0].numpy()

def show_cam(img, cam, ax, title):
    heat = cm.jet(tf.image.resize(cam[..., None], (224, 224)).numpy()[..., 0])[..., :3]
    ax.imshow(np.clip(0.55 * img / 255.0 + 0.45 * heat, 0, 1))
    ax.set_title(title, fontsize=8); ax.axis("off")
````

````python
def to_tflite(m, mode):
    c = tf.lite.TFLiteConverter.from_keras_model(m)
    if mode == "dynamic":
        c.optimizations = [tf.lite.Optimize.DEFAULT]
    if mode == "fp16":
        c.optimizations = [tf.lite.Optimize.DEFAULT]
        c.target_spec.supported_types = [tf.float16]
    return c.convert()                       # fp32: koi flag nahi

def eval_tflite(tfl):
    interp = tf.lite.Interpreter(model_content=tfl); interp.allocate_tensors()
    inp, out = interp.get_input_details()[0], interp.get_output_details()[0]
    def run(x):
        interp.set_tensor(inp["index"], x); interp.invoke()
        return interp.get_tensor(out["index"])[0]
    ok = n = 0
    for ims, lbs in test_ds:
        for im, lb in zip(ims.numpy(), lbs.numpy()):
            ok += int(run(im[None]).argmax() == lb); n += 1
    x = np.random.rand(1, 224, 224, 3).astype("float32") * 255
    for _ in range(10): run(x)
    t = []
    for _ in range(100):
        s = time.perf_counter(); run(x); t.append((time.perf_counter() - s) * 1000)
    return round(len(tfl)/1e6, 2), round(ok/n, 4), round(np.mean(t), 1), round(np.percentile(t, 95), 1)

for name, m in [("effnet", model2), ("mobilenet", model)]:
    for mode in ["fp32", "fp16", "dynamic"]:
        print(name, mode, eval_tflite(to_tflite(m, mode)))   # (MB, acc, ms mean, ms p95)
````
