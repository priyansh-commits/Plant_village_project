````python

# ============================================
# PLANT DISEASE CLASSIFICATION - COMPLETE CODE
# One MobileNetV2 model for all 15 classes
# ============================================

!pip -q install kagglehub

import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import kagglehub

from tensorflow.keras import layers, Model
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from sklearn.metrics import classification_report, ConfusionMatrixDisplay

# 1. DOWNLOAD DATASET
path = kagglehub.dataset_download("arjuntejaswi/plant-village")

# Automatically find the folder containing all disease classes
data_dir = os.path.join(path, "PlantVillage")

if not os.path.isdir(data_dir):
    raise FileNotFoundError(f"Dataset folder not found: {data_dir}")

class_names = sorted([
    d for d in os.listdir(data_dir)
    if os.path.isdir(os.path.join(data_dir, d))
])

assert len(class_names) == 15, (
    f"Expected 15 classes, found {len(class_names)}: {class_names}"
)

print("Dataset:", data_dir)
print("Number of classes:", len(class_names))
print(class_names)

# 2. LOAD DATA

import os
import numpy as np
import tensorflow as tf
from sklearn.model_selection import train_test_split
from collections import Counter

# -------------------------------
# 1. Configuration
# -------------------------------
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

# Supported image formats
IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp")

# -------------------------------
# 2. Read all classes and images
# -------------------------------
class_names = sorted([
    name for name in os.listdir(data_dir)
    if os.path.isdir(os.path.join(data_dir, name))
])

assert len(class_names) == 15, (
    f"Expected 15 classes, found {len(class_names)}"
)

image_paths = []
image_labels = []

for label, class_name in enumerate(class_names):
    class_dir = os.path.join(data_dir, class_name)

    files = sorted(os.listdir(class_dir))
    valid_files = [
        f for f in files
        if f.lower().endswith(IMAGE_EXTENSIONS)
        and os.path.isfile(os.path.join(class_dir, f))
    ]

    for filename in valid_files:
        image_paths.append(os.path.join(class_dir, filename))
        image_labels.append(label)

image_paths = np.array(image_paths)
image_labels = np.array(image_labels, dtype=np.int32)

print("Total images:", len(image_paths))
print("Total classes:", len(class_names))

# -------------------------------
# 3. Stratified train/test split
#    70% train, 30% temporary
# -------------------------------
train_paths, temp_paths, train_labels, temp_labels = (
    train_test_split(
        image_paths,
        image_labels,
        test_size=0.30,
        random_state=SEED,
        stratify=image_labels
    )
)

# -------------------------------
# 4. Stratified validation/test split
#    Split temporary set equally:
#    15% validation, 15% test
# -------------------------------
val_paths, test_paths, val_labels, test_labels = (
    train_test_split(
        temp_paths,
        temp_labels,
        test_size=0.50,
        random_state=SEED,
        stratify=temp_labels
    )
)

print("\nSplit sizes:")
print(f"Train:      {len(train_paths)} ({len(train_paths)/len(image_paths):.1%})")
print(f"Validation: {len(val_paths)} ({len(val_paths)/len(image_paths):.1%})")
print(f"Test:       {len(test_paths)} ({len(test_paths)/len(image_paths):.1%})")

# -------------------------------
# 5. Check per-class distribution
# -------------------------------
print("\nPer-class image counts:")

for i, name in enumerate(class_names):
    print(
        f"{i:2d} {name:48s}"
        f" Train={np.sum(train_labels == i):4d}"
        f" Val={np.sum(val_labels == i):4d}"
        f" Test={np.sum(test_labels == i):4d}"
    )

# -------------------------------
# 6. Image loading function
# -------------------------------
def load_image(path, label):
    image = tf.io.read_file(path)
    image = tf.io.decode_image(
        image,
        channels=3,
        expand_animations=False
    )
    image.set_shape([None, None, 3])
    image = tf.image.resize(image, IMG_SIZE)

    # Keep pixels in [0, 255].
    # MobileNetV2 preprocessing can remain inside your model.
    return image, label

# -------------------------------
# 7. Build datasets WITHOUT shuffle
# -------------------------------
def make_dataset(paths, labels):
    ds = tf.data.Dataset.from_tensor_slices((paths, labels))

    ds = ds.map(
        load_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    ds = ds.batch(BATCH_SIZE, drop_remainder=False)
    ds = ds.prefetch(tf.data.AUTOTUNE)

    return ds

train_ds = make_dataset(train_paths, train_labels)
val_ds = make_dataset(val_paths, val_labels)
test_ds = make_dataset(test_paths, test_labels)

print("\nDatasets created successfully!")
print("Train batches:", tf.data.experimental.cardinality(train_ds).numpy())
print("Validation batches:", tf.data.experimental.cardinality(val_ds).numpy())
print("Test batches:", tf.data.experimental.cardinality(test_ds).numpy())




# class_names = train_ds.class_names
NUM_CLASSES = len(class_names)

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)

# 3. DATA AUGMENTATION
augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.08),
    layers.RandomZoom(0.10),
])

# 4. LOAD PRETRAINED MOBILENETV2
base_model = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

base_model.trainable = False

inputs = layers.Input(shape=(224, 224, 3))

x = augmentation(inputs)
x = preprocess_input(x)

x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.30)(x)
outputs = layers.Dense(
    NUM_CLASSES,
    activation="softmax"
)(x)

model = Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# 5. TRAIN CLASSIFICATION HEAD
callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True
    ),
    tf.keras.callbacks.ModelCheckpoint(
        "/content/best_plant_model.keras",
        monitor="val_loss",
        save_best_only=True
    )
]

history1 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    callbacks=callbacks
)

# 6. FINE-TUNE LAST LAYERS OF MOBILENETV2
base_model.trainable = True

# Freeze most layers; fine-tune only the last 30
for layer in base_model.layers[:-30]:
    layer.trainable = False

# Keep BatchNormalization layers frozen for stable fine-tuning
for layer in base_model.layers:
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history2 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5,
    callbacks=callbacks
)

# 7. LOAD BEST SAVED MODEL
model = tf.keras.models.load_model(
    "/content/best_plant_model.keras"
)

# 8. EVALUATE
val_loss, val_accuracy = model.evaluate(val_ds, verbose=1)

print(f"\nValidation loss: {val_loss:.4f}")
print(f"Validation accuracy: {val_accuracy * 100:.2f}%")

# 9. PLOT TRAINING CURVES
acc = history1.history["accuracy"] + history2.history["accuracy"]
val_acc = (
    history1.history["val_accuracy"]
    + history2.history["val_accuracy"]
)

loss = history1.history["loss"] + history2.history["loss"]
val_loss_history = (
    history1.history["val_loss"]
    + history2.history["val_loss"]
)

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(acc, label="Training accuracy")
plt.plot(val_acc, label="Validation accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.title("Accuracy")

plt.subplot(1, 2, 2)
plt.plot(loss, label="Training loss")
plt.plot(val_loss_history, label="Validation loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.title("Loss")

plt.tight_layout()
plt.show()

# 10. CLASSIFICATION REPORT
y_true = np.concatenate([
    labels.numpy() for _, labels in val_ds
])

y_prob = model.predict(val_ds, verbose=1)
y_pred = np.argmax(y_prob, axis=1)

print(classification_report(
    y_true,
    y_pred,
    labels=np.arange(NUM_CLASSES),
    target_names=class_names,
    zero_division=0
))

# 11. CONFUSION MATRIX
plt.figure(figsize=(15, 15))

ConfusionMatrixDisplay.from_predictions(
    y_true,
    y_pred,
    labels=np.arange(NUM_CLASSES),
    display_labels=class_names,
    xticks_rotation=90,
    cmap="Blues",
    normalize="true",
    values_format=".2f"
)

plt.title("Normalized Confusion Matrix")
plt.tight_layout()
plt.show()

# 12. SAVE FINAL MODEL AND CLASS LABELS
model.save("/content/plant_disease_model.keras")

with open("/content/plant_class_names.txt", "w") as f:
    f.write("\n".join(class_names))

print("Model saved: /content/plant_disease_model.keras")
print("Class labels saved: /content/plant_class_names.txt")

# 13. TEST ON ONE VALIDATION IMAGE
for images, labels in val_ds.take(1):
    image = images[0:1]
    actual_label = int(labels[0].numpy())

    probabilities = model.predict(image, verbose=0)[0]
    predicted_label = int(np.argmax(probabilities))

    plt.figure(figsize=(5, 5))
    plt.imshow(images[0].numpy().astype("uint8"))
    plt.axis("off")
    plt.title(
        f"Predicted: {class_names[predicted_label]}\n"
        f"Confidence: {probabilities[predicted_label] * 100:.2f}%\n"
        f"Actual: {class_names[actual_label]}"
    )
    plt.show()

    print("Top 3 predictions:")
    for idx in np.argsort(probabilities)[-3:][::-1]:
        print(f"{class_names[idx]}: {probabilities[idx] * 100:.2f}%")
    break
````

````python
import keras
from keras import layers
from keras.optimizers import Adamax
from keras.callbacks import EarlyStopping

# Load pretrained EfficientNetB0 WITHOUT ImageNet classifier
base_model2 = keras.applications.EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(224, 224, 3)
)

# Stage 1: Freeze pretrained layers
base_model2.trainable = False

# Build classifier
model2 = keras.Sequential([
    keras.Input(shape=(224, 224, 3)),
    base_model2,
    layers.GlobalAveragePooling2D(),
    layers.BatchNormalization(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.3),
    layers.Dense(15, activation="softmax")
])

# Compile
model2.compile(
    optimizer=Adamax(learning_rate=0.0003),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model2.summary()
````

````python
early_stop = EarlyStopping(
    monitor='val_accuracy',
    patience=5,
    restore_best_weights=True
)
history = model2.fit(
    train_ds,
    validation_data=val_ds,
    epochs=20,
    callbacks=[early_stop],
    verbose=1
)
````
